```python
import random

def generate_mem_files(count=1024):
    amem = []
    bmem = []
    cmem = []
    
    for i in range(count):
        # -128 ~ +127 범위의 랜덤 정수 생성
        a_val = random.randint(-128, 127)
        b_val = random.randint(-128, 127)
        
        # 2의 보수 덧셈 (16비트 결과)
        c_val = a_val + b_val
        
        amem.append(a_val & 0xFF)
        bmem.append(b_val & 0xFF)
        cmem.append(c_val & 0xFFFF)
    
    # amem_1024.mem 생성
    with open("amem_1024.mem", 'w') as f:
        for val in amem:
            f.write(f"{val:02X}\n")
    
    # bmem_1024.mem 생성
    with open("bmem_1024.mem", 'w') as f:
        for val in bmem:
            f.write(f"{val:02X}\n")
    
    # cmem_1024.mem 생성 (16비트)
    with open("cmem_1024.mem", 'w') as f:
        for val in cmem:
            f.write(f"{val:04X}\n")

generate_mem_files()
print("amem_1024.mem, bmem_1024.mem, cmem_1024.mem 파일 생성 완료!")
```

```python
import random

def generate_coe_files(count=1024):
    amem = []
    bmem = []
    cmem = []
    
    for i in range(count):
        # -128 ~ +127 범위의 랜덤 정수 생성
        a_val = random.randint(-128, 127)
        b_val = random.randint(-128, 127)
        
        # 2의 보수 덧셈 (16비트 결과)
        c_val = a_val + b_val
        
        amem.append(a_val & 0xFF)
        bmem.append(b_val & 0xFF)
        cmem.append(c_val & 0xFFFF)
    
    # amem_1024.coe 생성 (8비트)
    with open("amem_1024.coe", 'w') as f:
        f.write("memory_initialization_radix=16;\n")
        f.write("memory_initialization_vector=\n")
        for i, val in enumerate(amem):
            if i < len(amem) - 1:
                f.write(f"{val:02X},\n")
            else:
                f.write(f"{val:02X};\n")
    
    # bmem_1024.coe 생성 (8비트)
    with open("bmem_1024.coe", 'w') as f:
        f.write("memory_initialization_radix=16;\n")
        f.write("memory_initialization_vector=\n")
        for i, val in enumerate(bmem):
            if i < len(bmem) - 1:
                f.write(f"{val:02X},\n")
            else:
                f.write(f"{val:02X};\n")
    
    # cmem_1024.coe 생성 (16비트)
    with open("cmem_1024.coe", 'w') as f:
        f.write("memory_initialization_radix=16;\n")
        f.write("memory_initialization_vector=\n")
        for i, val in enumerate(cmem):
            if i < len(cmem) - 1:
                f.write(f"{val:04X},\n")
            else:
                f.write(f"{val:04X};\n")

generate_coe_files()
print("amem_1024.coe, bmem_1024.coe, cmem_1024.coe 파일 생성 완료!")
```
