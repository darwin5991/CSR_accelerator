```python
import matplotlib.pyplot as plt
import numpy as np
from torch import nn
from torch import optim
import torch
```

```python
N = 100 # number of points per class
D = 2 # dimensionality
K = 3 # number of classes
device = torch.device('cpu')  # 또는 'cuda'
```

```python
# X = np.zeros((N*K,D)) # data matrix (each row = single example)
# y = np.zeros(N*K, dtype='uint8') # class labels

# for j in range(K): # for each class
#   ix = range(N*j,N*(j+1))
#   r = np.linspace(0.0,1,N) # radius
#   t = np.linspace(j*(4.5),(j+1)*(4.5),N) + np.random.randn(N)*0.2 # theta
#   X[ix] = np.c_[r*np.sin(t), r*np.cos(t)]
#   y[ix] = j

# # 변환 방법 1: from_numpy (메모리 공유, dtype 맞춰야 함)
# X_t = torch.from_numpy(X).float()      # float32로 변환
# y_t = torch.from_numpy(y).long()       # 정수 레이블은 long

# # 변환 방법 2: torch.tensor (항상 복사)
# X_t2 = torch.tensor(X, dtype=torch.float32)
# y_t2 = torch.tensor(y, dtype=torch.long)
```

```python
# X = torch.zeros((N*K, D), dtype=torch.float32, device=device)
# y = torch.zeros(N*K, dtype=torch.long, device=device)

# for j in range(K):
#     ix = slice(N*j, N*(j+1))
#     r = torch.linspace(0.0, 1.0, N, device=device)
#     t = torch.linspace(j*4.5, (j+1)*4.5, N, device=device) + torch.randn(N, device=device)*0.2
#     X[ix] = torch.stack([r * torch.sin(t), r * torch.cos(t)], dim=1)
#     y[ix] = j
```

```python
r = torch.linspace(0.0,1.0,N,device=device).repeat(K)                 # (N*K,)
print(r.shape)
t = torch.cat([torch.linspace(j*4.5,(j+1)*4.5,N,device=device) for j in range(K)]) \
    + torch.randn(N*K, device=device)*0.2


X = torch.stack([r * torch.sin(t), r * torch.cos(t)], dim=1)          # (N*K,2)
y = torch.arange(K, device=device).repeat_interleave(N)               # (N*K,)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y, s=40, cmap=plt.cm.Spectral)
plt.show()
```

```python
#linear
model=nn.Linear(2,3)
print(list(model.parameters()))
#하이퍼 파라미터 설정
LR=1e-0
EPOCH=200
criterion=nn.CrossEntropyLoss()

loss_history=[]

for ep in range(EPOCH):
    #forwardpath
    y_pred=model(X)

    loss=criterion(y_pred, y)
    loss_with_reg=loss+0.01*torch.sum(model.weight**2)

    # print(loss.type())

    #backprop
    loss_with_reg.backward()
    #update
    with torch.no_grad():
        for param in model.parameters():
            param -= LR * param.grad
        model.zero_grad()

    #print loss
    loss_history+=[loss_with_reg.item()]
    if ep % 20 == 0:
        print(f'Epoch {ep+1}/{EPOCH}, Loss: {loss_with_reg.item():.4f}')
        #print weight and bias
        print(f"weight:{model.weight}, bias:{model.bias}")
        #결정경계 plot
        with torch.no_grad():
            x_min, x_max = X[:,0].min().item() - 0.5, X[:,0].max().item() + 0.5
            y_min, y_max = X[:,1].min().item() - 0.5, X[:,1].max().item() + 0.5
            xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200), np.linspace(y_min, y_max, 200))
            grid = torch.from_numpy(np.c_[xx.ravel(), yy.ravel()]).float().to(device)
            Z = model(grid)
            Z = Z.argmax(dim=1).cpu().numpy().reshape(xx.shape)
            plt.figure(figsize=(5,4))
            plt.contourf(xx, yy, Z, cmap=plt.cm.Spectral, alpha=0.6)
            plt.scatter(X[:,0].cpu().numpy(), X[:,1].cpu().numpy(), c=y.cpu().numpy(), s=40, cmap=plt.cm.Spectral, edgecolors='k')
            plt.title(f'Epoch {ep+1}')
            plt.show()

plt.plot(loss_history)
```

```python
#NN
model=nn.Sequential(nn.Linear(2,100),
                    nn.ReLU(),
                    nn.Linear(100,3)
                    )
# print(list(model[0].parameters()))
#하이퍼 파라미터 설정
LR=1e0
REG=1e-3
EPOCH=10000
criterion=nn.CrossEntropyLoss()

loss_history=[]
acc_history=[]

for ep in range(EPOCH):
    #forwardpath
    y_pred=model(X)

    # loss=criterion(y_pred, y)
    # loss_with_reg=loss+REG*torch.stack([ (p**2).sum() for p in model.parameters() if p.requires_grad and p.dim() > 1 ]).sum()  # bias 제외
    ce = criterion(y_pred, y)
    l2 = torch.stack([(p**2).sum() for p in model.parameters() if p.requires_grad and p.dim() > 1]).sum()
    loss = ce + 0.5*REG * l2
    # print(loss.type())

    #backprop
    loss.backward()
    #update
    with torch.no_grad():
        for param in model.parameters():
            param -= LR * param.grad
        model.zero_grad()

    #print loss
    loss_history+=[loss.item()]
    preds = y_pred.argmax(dim=1)
    acc = (preds == y).float().mean().item()
    acc_history.append(acc)
    if ep % 2500 ==2499:
        print(f'Epoch {ep+1}/{EPOCH}, Loss: {loss.item():.4f}, Accuracy: {acc:.4f}')
        #print weight and bias
        # print(f"weight:{model.weight}, bias:{model.bias}")
        #결정경계 plot
        with torch.no_grad():
            x_min, x_max = X[:,0].min().item() - 0.5, X[:,0].max().item() + 0.5
            y_min, y_max = X[:,1].min().item() - 0.5, X[:,1].max().item() + 0.5
            xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200), np.linspace(y_min, y_max, 200))
            grid = torch.from_numpy(np.c_[xx.ravel(), yy.ravel()]).float().to(device)
            Z = model(grid)
            Z = Z.argmax(dim=1).cpu().numpy().reshape(xx.shape)
            plt.figure(figsize=(5,4))
            plt.contourf(xx, yy, Z, cmap=plt.cm.Spectral, alpha=0.6)
            plt.scatter(X[:,0].cpu().numpy(), X[:,1].cpu().numpy(), c=y.cpu().numpy(), s=40, cmap=plt.cm.Spectral, edgecolors='k')
            plt.title(f'Epoch {ep+1}')
            plt.show()

plt.plot(loss_history)
plt.plot(acc_history)
```

```python
class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        # self.conv1=nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, stride=1, padding=1)
        # self.maxPool=nn.MaxPool2d(kernel_size=2, stride=2)
        # self.conv2=nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, stride=1, padding=1)
        self.conv_layers=nn.Sequential( nn.Conv2d(3,64,3,padding=1),
                                        nn.BatchNorm2d(64),
                                        nn.ReLU(),
                                        nn.Conv2d(64,64,3,padding=1),
                                        nn.BatchNorm2d(64),
                                        nn.ReLU(),
                                        nn.MaxPool2d(2), # 32x32 -> 16x16

                                        nn.Conv2d(64,128,3,padding=1),
                                        nn.BatchNorm2d(128),
                                        nn.ReLU(),
                                        nn.Conv2d(128,128,3,padding=1),
                                        nn.BatchNorm2d(128),
                                        nn.ReLU(),
                                        nn.MaxPool2d(2), # 16x16 -> 8x8

                                        nn.Conv2d(128,256,3,padding=1),
                                        nn.BatchNorm2d(256),
                                        nn.ReLU(),
                                        nn.Conv2d(256,256,3,padding=1),
                                        nn.BatchNorm2d(256),
                                        nn.ReLU(),
                                        nn.Conv2d(256,256,3,padding=1),
                                        nn.BatchNorm2d(256),
                                        nn.ReLU(),
                                        nn.MaxPool2d(2)) # 8x8 -> 4x4
        self.flat=nn.Flatten(start_dim=1)
        self.fc_layers=nn.Sequential( nn.Linear(256*4*4,1024),
                                      nn.ReLU(),
                                      nn.Dropout(0.5),
                                      nn.Linear(1024,256),
                                      nn.ReLU(),
                                      nn.Dropout(0.5),
                                      nn.Linear(256,10))

    def forward(self, x):
        x=self.conv_layers(x)
        x=self.flat(x)
        x=self.fc_layers(x)
        return x
```

```python
def Train(model, train_DL, EPOCH, criterion, optimizer):
    loss_history=[]
    acc_history=[]
    NOT=len(train_DL.dataset)
    model.train()
    check_ep=EPOCH//10
    for ep in range(EPOCH):
        running_loss=0
        running_acc=0
        for x_batch, y_batch in train_DL:
            #forward
            x_batch= x_batch.to(DEVICE)
            y_batch= y_batch.to(DEVICE)
            y_pred=model(x_batch)
            batch_loss=criterion(y_pred, y_batch)
            #backprop
            optimizer.zero_grad()
            batch_loss.backward()
            #update
            optimizer.step()
            running_loss+= batch_loss.item() * x_batch.size(0)
            running_acc+= (y_pred.argmax(dim=1)==y_batch).sum().item()

        ep_loss= running_loss / NOT
        ep_acc= running_acc / NOT
        loss_history.append(ep_loss)
        acc_history.append(ep_acc)
        if (ep+1) % check_ep ==0:
            print(f"Epoch: {ep+1}/{EPOCH}, Loss: {ep_loss:.4f}, Accuracy: {ep_acc*100:.2f} %")

    return loss_history, acc_history
```
